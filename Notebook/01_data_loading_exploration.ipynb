```python
from pathlib import Path

# All notebooks anchor paths off THIS notebook's own folder, not the
# current working directory -- this way it doesn't matter whether you
# launch Jupyter from the project root, from notebooks/, or from an IDE.
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA = ROOT / "data" / "processed"
MODELS = ROOT / "models"
DATA.mkdir(parents=True, exist_ok=True)
MODELS.mkdir(parents=True, exist_ok=True)
print("Project root:", ROOT)
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/SnehaDeshmukh28/IndianBailJudgments-1200/main/indian_bail_judgments.csv"
df = pd.read_csv(url)

print("Shape:", df.shape)
df.head()
```

```python

print(df.columns.tolist())
```

```python
print(df['crime_type'].value_counts())
```

```python
target_categories = ['Theft or Robbery', 'Cyber Crime', 'Fraud or Cheating']

filtered_df = df[df['crime_type'].isin(target_categories)].copy()

print("Total filtered cases:", filtered_df.shape)
print(filtered_df['crime_type'].value_counts())
```

```python
import os


os.makedirs(DATA, exist_ok=True)
filtered_df.to_csv(DATA / "filtered_low_severity_cases.csv", index=False)
print("Saved:", DATA / "filtered_low_severity_cases.csv", filtered_df.shape)
```
